**Logistic Regression for Heart Disease Prediction**

Objective

To build a Logistic Regression model that predicts whether a person has heart disease based on various medical attributes.

About the Dataset

The Heart Disease dataset (heart.csv) contains medical information about individuals, such as age, sex, chest pain type, blood pressure, cholesterol, and other health-related attributes. The target column is the dependent variable, which indicates whether heart disease is present.

* Target = 0: No heart disease
* Target = 1: Heart disease present

```python
# Step 1: Import Libraries

import pandas as pd

import matplotlib.pyplot as plt

import seaborn as sns

from sklearn.model_selection import train_test_split

from sklearn.linear_model import LogisticRegression

from sklearn.metrics import accuracy_score, confusion_matrix
```

```python
# Step 2: Import Dataset and Explore
df= pd.read_csv('/heart.csv')
print(df.describe())
df.info()
```

```python
# Step 3: Remove Unnecessary Columns
df.drop_duplicates(inplace=True)
sns.countplot(data=df, x='target')
plt.title("Count Plot for target variable")
plt.show()
```

```python
# Step 4: Separate Independent and Dependent Variables
X=df.drop('target',axis=1)
y=df['target']
```

```python
# Step 5: Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(

    X, y, test_size=0.2, random_state=1

)
```

```python
# Step 6: Model Training
model = LogisticRegression(max_iter=1000)

model.fit(X_train, y_train)
```

```python
# Step 7: Prediction
y_pred=model.predict(y_test)
print("Predicted Values:", y_pred)
```

```python
# Step 8: Model Evaluation

print("Accuracy:", accuracy_score(y_test, y_pred))

cm = confusion_matrix(y_test, y_pred)

print(cm)
```

```python
# Step 9: Visualization

sns.heatmap(cm, annot=True, fmt='d')

plt.xlabel("Predicted")

plt.ylabel("Actual")

plt.show()
```
